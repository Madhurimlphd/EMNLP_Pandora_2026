Calculated Fleiss kappa between human annotators and for <br/> [1] female persona and <br/> [2] male persona <br/> [3] Age1 persona <br/>[4] Age2 persona <br/> [5] Age3 persona

```python
import pandas as pd
#df = pd.read_csv("LLM_annotation2/All_LLM_Prompt2_Combined.csv")
#df = pd.read_csv("LLM_annotation3_female_persona/Prompt3_female_persona.csv")
df = pd.read_csv("LLM_annotation4_male_persona/Prompt4_male_persona.csv")
```

```python
from sklearn.metrics import cohen_kappa_score
import itertools
```

```python
#female annotators
#label_cols = ["anno1", "anno2", "anno6","GPT_5.4_nano","GPT_5.4","GPT_5.4_mini","GPT_5.5","phi","mistral"]

#male annotators
label_cols = ["anno3", "anno4", "anno5","GPT_5.4_nano","GPT_5.4","GPT_5.4_mini","GPT_5.5","phi","mistral"]

# compute pairwise kappa
for col1, col2 in itertools.combinations(label_cols, 2):
    kappa = cohen_kappa_score(df[col1], df[col2])
    print(f"{col1} vs {col2}: {kappa:.4f}")
```

**Female persona**

```python
from statsmodels.stats.inter_rater import aggregate_raters,fleiss_kappa



df = pd.read_csv("LLM_annotation3_female_persona/Prompt3_female_persona.csv")

#female annotators
df_ann = df[["anno1", "anno2", "anno6"]]



df_ann = df_ann.apply(lambda col: col.map({"YES": 1, "NO": 0})).astype(int)

table, _ = aggregate_raters(df_ann.to_numpy()) 

fleiss = fleiss_kappa(table)
print("Fleiss kappa for the female annotators")
print(fleiss)
```

```python
#female annotators
df_closed_llm = df[["anno1", "anno2", "anno6","GPT_5.4_nano","GPT_5.4","GPT_5.4_mini","GPT_5.5"]]



df_closed_llm = df_closed_llm.apply(lambda col: col.map({"YES": 1, "NO": 0})).astype(int)

table, _ = aggregate_raters(df_closed_llm.to_numpy()) 

fleiss = fleiss_kappa(table)

print("Fleiss kappa for female annotators and closed LLM models ")
print(fleiss)
```

```python
#female annotators
df_open_llm = df[["anno1", "anno2", "anno6","phi","mistral"]]


df_open_llm = df_open_llm.apply(lambda col: col.map({"YES": 1, "NO": 0})).astype(int)

table, _ = aggregate_raters(df_open_llm.to_numpy()) 

fleiss = fleiss_kappa(table)

print("Fleiss kappa for female annotators and open LLM models  ")
print(fleiss)
```

**Age1 persona**

```python
from statsmodels.stats.inter_rater import aggregate_raters,fleiss_kappa
import ast 


df1 = pd.read_csv("LLM_annotation5_age_persona/GPT_5.4_EXIST2023_age18.csv")
df2 = pd.read_csv("LLM_annotation5_age_persona/GPT_5.4_mini_EXIST2023_age18.csv")
df3 = pd.read_csv("LLM_annotation5_age_persona/GPT_5.4_nano_EXIST2023_age18.csv")
df4 = pd.read_csv("LLM_annotation5_age_persona/GPT_5.5_EXIST2023_age18.csv")
df5 = pd.read_csv("LLM_annotation5_age_persona/phi_EXIST2023_age18.csv")
df6 = pd.read_csv("LLM_annotation5_age_persona/mistral_EXIST2023_age18.csv")


df_llm = (
    df1
    .merge(df2[["id_EXIST", "GPT_5.4_mini"]], on="id_EXIST", how="left")
    .merge(df3[["id_EXIST", "GPT_5.4_nano"]], on="id_EXIST", how="left")
    .merge(df4[["id_EXIST", "GPT_5.5"]], on="id_EXIST", how="left")
    .merge(df5[["id_EXIST", "phi"]], on="id_EXIST", how="left")
    .merge(df6[["id_EXIST", "mistral"]], on="id_EXIST", how="left")
)

# Convert string → list
df_llm["labels_task1"] = df_llm["labels_task1"].apply(ast.literal_eval)

# Split into 6 columns
labels_df = pd.DataFrame(df_llm["labels_task1"].tolist(),columns=["anno1", "anno2", "anno3", "anno4", "anno5", "anno6"])

# Merge back
df_llm = pd.concat([df_llm, labels_df], axis=1)

llm =df_llm[["anno1","anno3"]]


llm = llm.apply(lambda col: col.map({"YES": 1, "NO": 0})).astype(int)

table, _ = aggregate_raters(llm.to_numpy()) 

fleiss = fleiss_kappa(table)
print("Fleiss kappa for the age1 human annotators")
print(fleiss)
```

```python
llm =df_llm[["anno1","anno3","GPT_5.4","GPT_5.4_mini","GPT_5.4_nano","GPT_5.5"]]


llm = llm.apply(lambda col: col.map({"YES": 1, "NO": 0})).astype(int)

table, _ = aggregate_raters(llm.to_numpy()) 

fleiss = fleiss_kappa(table)
print("Fleiss kappa for the human+GPT age1 persona models")
print(fleiss)
```

```python
llm =df_llm[["anno1","anno3","phi","mistral"]]


llm = llm.apply(lambda col: col.map({"YES": 1, "NO": 0})).astype(int)

table, _ = aggregate_raters(llm.to_numpy())

fleiss = fleiss_kappa(table)
print("Fleiss kappa for the human+Open LLM age1 persona models")
print(fleiss)
```

```python
llm =df_llm[["anno1","anno3","GPT_5.4","GPT_5.4_mini","GPT_5.4_nano","GPT_5.5","phi","mistral"]]


llm = llm.apply(lambda col: col.map({"YES": 1, "NO": 0})).astype(int)

table, _ = aggregate_raters(llm.to_numpy()) 

fleiss = fleiss_kappa(table)
print("Fleiss kappa for the human+All LLM age1 persona models")
print(fleiss)
```
