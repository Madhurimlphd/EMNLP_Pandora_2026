In this notebook, <br/> [1] Cohen Kappa between individual human annotations and LLM models <br/> [2] Fleiss kappa between individual human annotations and LLM models <br/> [3] Krippendorff' alpha between individual human annotations and LLM models <br/> [4] Cohen Kappa between label aggregation strategies (majority, dawid skene and mace) and LLM models <br/>

```python
import pandas as pd
df = pd.read_csv("LLM_annotation2/All_LLM_Prompt2_Combined.csv")
```

```python
from sklearn.metrics import cohen_kappa_score
import itertools
```

```python
label_cols = ["anno1", "anno2", "anno3", "anno4", "anno5", "anno6","GPT_5.4_nano","GPT_5.4","GPT_5.4_mini","GPT_5.5","mistral","phi"]



# compute pairwise kappa
for col1, col2 in itertools.combinations(label_cols, 2):
    kappa = cohen_kappa_score(df[col1], df[col2])
    print(f"{col1} vs {col2}: {kappa:.4f}")
```

```python
from statsmodels.stats.inter_rater import aggregate_raters,fleiss_kappa



#df = pd.read_csv("LLM_processed_dataset/All_LLM_EXIST2023.csv")

df = pd.read_csv("LLM_annotation2/All_LLM_Prompt2_Combined.csv")
df_ann =df[["anno1", "anno2", "anno3", "anno4", "anno5", "anno6"]]
df_ann = df_ann.apply(lambda col: col.map({"YES": 1, "NO": 0})).astype(int)

table, _ = aggregate_raters(df_ann.to_numpy()) #for fleiss kappa, we need to aggregated annotators; rows as subjects and annotators as columns; fleiss kappa consider (items, annotators)
#print(table)

fleiss = fleiss_kappa(table)
print("Fleiss kappa for the human annotators")
print(fleiss)
```

**Cohen Kappa between three label aggregation strategies**

```python
k1=cohen_kappa_score(df["majority_human"].map({"YES": 1, "NO": 0}),df["ds_human"])
k2=cohen_kappa_score(df["majority_human"].map({"YES": 1, "NO": 0}),df["mc_human"])
k3=cohen_kappa_score(df["ds_human"],df["mc_human"])
print(f"Majority vs Dawid Skene: {k1:.4f}")
print(f"Majority vs MACE: {k2:.4f}")
print(f"Dawid Skene vs MACE: {k3:.4f}")
```

**Cohen Kappa between label aggregation strategies (majority, dawid skene and mace) and GPT models**

```python
#from sklearn.metrics import cohen_kappa_score
kappa_score_mg=cohen_kappa_score(df["majority_human"].map({"YES": 1, "NO": 0}),df["GPT_5.4_nano"].map({"YES": 1, "NO": 0}))
kappa_score_dg=cohen_kappa_score(df["ds_human"],df["GPT_5.4_nano"].map({"YES": 1, "NO": 0}))
kappa_score_cg=cohen_kappa_score(df["mc_human"],df["GPT_5.4_nano"].map({"YES": 1, "NO": 0}))
print("===== Cohen Kappa Scores =====")
print(f"Human Majority Vote vs GPT_5.4_nano     : {kappa_score_mg:.4f}")
print(f"Dawid-Skene vs GPT_5.4_nano             : {kappa_score_dg:.4f}")
print(f"MACE vs GPT_5.4_nano                    : {kappa_score_cg:.4f}")
```
