**Calculated the Jensen–Shannon divergence between human annotators and the LLMs, with each LLM annotated using all six prompts.**

**Human vs GPT 5.4**

```python
from scipy.spatial.distance import jensenshannon #this function gives Jensen-Shannon distance between two probability arrays. this is the square root of the Jensen-Shannon divergence


def jsd(df_llm,llm_cols,anno_cols):

    df_llm["llm_prob"] = (
        df_llm[llm_cols]
        .apply(lambda row: row.astype(str).str.upper().eq("YES").mean(), axis=1)
    )

    df_llm["human_prob"] = (
        df_llm[anno_cols]
        .apply(lambda row: row.astype(str).str.upper().eq("YES").mean(), axis=1)
    )

    # Round human_prob to 2 decimal places
    df_llm["human_prob"] = df_llm["human_prob"].round(2)

    # If you also want to round llm_prob
    df_llm["llm_prob"] = df_llm["llm_prob"].round(2)
    
    df_llm["jsd"] = df_llm.apply(
        lambda row: jensenshannon([row["human_prob"], 1-row["human_prob"]],[row["llm_prob"], 1-row["llm_prob"]],base=2
        #),  #this gives Jensen Shannon distance
        )**2,     # square it to obtain Jensen-Shannon divergence 
        axis=1
    )
    return df_llm
```

```python
import pandas as pd
df1 = pd.read_csv("LLM_annotation2/GPT_5.4_EXIST2023.csv")
df2=pd.read_csv("LLM_annotation3_female_persona/GPT_5.4_EXIST2023.csv")
df3=pd.read_csv("LLM_annotation4_male_persona/GPT_5.4_EXIST2023.csv")
df4=pd.read_csv("LLM_annotation5_age_persona/GPT_5.4_EXIST2023_age18.csv")
df5=pd.read_csv("LLM_annotation5_age_persona/GPT_5.4_EXIST2023_age23.csv")
df6=pd.read_csv("LLM_annotation5_age_persona/GPT_5.4_EXIST2023_age46.csv")


df=pd.read_csv("LLM_annotation2/All_LLM_Prompt2_Combined.csv")
```

```python


df_llm = (
    df1
    .merge(df1[["id_EXIST", "GPT_5.4"]], on="id_EXIST", how="left",suffixes=("", "_df1"))
    .merge(df2[["id_EXIST", "GPT_5.4"]], on="id_EXIST", how="left", suffixes=("", "_df2"))
    .merge(df3[["id_EXIST", "GPT_5.4"]], on="id_EXIST", how="left", suffixes=("", "_df3"))
    .merge(df4[["id_EXIST", "GPT_5.4"]], on="id_EXIST", how="left", suffixes=("", "_df4"))
    .merge(df5[["id_EXIST", "GPT_5.4"]], on="id_EXIST", how="left", suffixes=("", "_df5"))
    .merge(df6[["id_EXIST", "GPT_5.4"]], on="id_EXIST", how="left", suffixes=("", "_df6"))
)

df_llm = df_llm.merge(df[["id_EXIST","anno1", "anno2", "anno3", "anno4", "anno5", "anno6"]],on="id_EXIST",how="left")


llm_cols =["GPT_5.4_df1","GPT_5.4_df2","GPT_5.4_df3","GPT_5.4_df4","GPT_5.4_df5","GPT_5.4_df6"]
anno_cols = ["anno1", "anno2", "anno3", "anno4", "anno5", "anno6"]

df_llm=jsd(df_llm,llm_cols,anno_cols)
print(df_llm["jsd"].describe())
```
