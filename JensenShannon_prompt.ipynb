```python
prompt1_cols=["GPT_5.4","GPT_5.4_mini","GPT_5.4_nano","GPT_5.5","phi","mistral"]
prompt2_cols=["GPT_5.4_df2","GPT_5.4_mini_df2","GPT_5.4_nano_df2","GPT_5.5_df2","phi_df2","mistral_df2"]
prompt3_cols=["GPT_5.4_df3","GPT_5.4_mini_df3","GPT_5.4_nano_df3","GPT_5.5_df3","phi_df3","mistral_df3"]
prompt4_cols=["GPT_5.4_df4","GPT_5.4_mini_df4","GPT_5.4_nano_df4","GPT_5.5_df4","phi_df4","mistral_df4"]
prompt5_cols=["GPT_5.4_df5","GPT_5.4_mini_df5","GPT_5.4_nano_df5","GPT_5.5_df5","phi_df5","mistral_df5"]
prompt6_cols=["GPT_5.4_df6","GPT_5.4_mini_df6","GPT_5.4_nano_df6","GPT_5.5_df6","phi_df6","mistral_df6"]
```

```python
from scipy.spatial.distance import jensenshannon #this function gives Jensen-Shannon distance between two probability arrays. this is the square root of the Jensen-Shannon divergence


def jsd(df_llm,llm1_cols,llm2_cols):

    df_llm["llm1_prob"] = (
        df_llm[llm1_cols]
        .apply(lambda row: row.astype(str).str.upper().eq("YES").mean(), axis=1)
    )

    df_llm["llm2_prob"] = (
        df_llm[llm2_cols]
        .apply(lambda row: row.astype(str).str.upper().eq("YES").mean(), axis=1)
    )

    # Round human_prob to 2 decimal places
    df_llm["llm1_prob"] = df_llm["llm1_prob"].round(2)

    # If you also want to round llm_prob
    df_llm["llm2_prob"] = df_llm["llm2_prob"].round(2)
    
    df_llm["jsd"] = df_llm.apply(
        lambda row: jensenshannon([row["llm1_prob"], 1-row["llm1_prob"]],[row["llm2_prob"], 1-row["llm2_prob"]],base=2
        #),  #this gives Jensen Shannon distance
        )**2,     # square it to obtain Jensen-Shannon divergence 
        axis=1
    )
    return df_llm
```
