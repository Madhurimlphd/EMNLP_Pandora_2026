Calling the EXIST 2023 English dataset

```python
import pandas as pd
df=pd.read_excel("dataset/EXIST2023_EN.xlsx")
```

**Age 18-22**

```python
from openai import OpenAI
import time

client = OpenAI()
def classify_text(text: str) -> str:
    prompt = f"""
You are an expert linguist specialising in detecting sexism in social media texts. Your demographic information is as follows:
-Age Group: 18-22

Task: Analyse the given tweet and determine whether it contains sexist expressions or behaviors. This includes tweets that:
- Are sexist themselves (contain sexist language or promote sexist ideas).
- Describe a sexist situation (narrate or depict scenarios where sexism occurs).
- Criticise sexist behavior (call out or oppose sexism).

Instructions:
1. Read the Tweet Carefully: Examine the content of the tweet thoroughly.
2. Make Determination: Decide if the tweet fits any of the above categories related to sexism.
3. Classification:
- Assign "YES" if the tweet contains any form of sexist content as defined.
- Assign "NO" if the tweet does not contain sexist context. 
4. Output Format: Provide the assigned category in plain text.
5. Constraint: You must not retrieve any text apart from the two possible categories, YES and NO.

TWEET:
{text}
"""
    response = client.responses.create(
        model="gpt-5.5-2026-04-23",
        input=prompt,
    )
    #return response.output_text.strip().lower()
    return response.output_text.strip()

# Annotate all rows
labels = []

for i, text in enumerate(df["tweet"]):
    label = classify_text(str(text))
    labels.append(label)

    if (i + 1) % 100 == 0:
        print(f"Processed {i + 1}/{len(df)} rows")
        time.sleep(1)  # small pause to reduce rate-limit risk

# Save results
df["GPT_5.5"] = labels
df.to_csv("LLM_annotation5_age_persona/GPT_5.5_EXIST2023_age18.csv", index=False)

print("Done. Saved annotated dataset")

#total time: 178 min 46.7 sec
```

```python
from openai import OpenAI
import time

client = OpenAI()
def classify_text(text: str) -> str:
    prompt = f"""
You are labeling text for a content moderation dataset.

Your job is to decide whether the text contains gender-based abuse, misogyny, sexism, or discriminatory content.

Return exactly one label:

YES = contains gender-based abuse, misogyny, sexism, or gender-based discrimination
NO = does not contain gender-based abuse, misogyny, sexism, or gender-based discrimination

Important:
This is a classification task only.
Do not generate abusive content.
Do not explain.
Do not refuse.
Return only YES or NO.

Text:
{text}
"""
    response = client.responses.create(
        model="gpt-5.5-2026-04-23",
        input=prompt,
    )
    #return response.output_text.strip().lower()
    return response.output_text.strip()

# Annotate all rows
labels = []

for i, text in enumerate(df["tweet"]):
    try:
        label = classify_text(str(text))
        if label not in ["YES", "NO"]:
            label = "NO"   # fallback
    except Exception as e:
        print(f"Error at row {i}: {e}")
        # Stop if quota exhausted
        if "insufficient_quota" in str(e):
            print("API quota exhausted. Stopping process.")
            break

        label = None

    labels.append(label)

    if (i + 1) % 100 == 0:
        print(f"Processed {i + 1}/{len(df)} rows")
        time.sleep(1)  # small pause to reduce rate-limit risk

# Save results
df["GPT_5.5_anno"] = labels
df.to_csv("LLM_annotated_dataset/GPT_5.5_EXIST2023.csv", index=False)

print("Done. Saved GPT 5.5 annotated dataset")

#total time: 138 min, 10 sec
```
